```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense, Flatten
from mlxtend.plotting import plot_decision_regions
from sklearn.datasets import make_moons
from tensorflow.keras.optimizers import Adam
```

```python
X, Y = make_moons(n_samples=100, noise=0.25, random_state=42)
```

```python
sns.scatterplot(x = X[:, 0], y = X[:, 1], c = Y)
```

#Without Regularization

```python
model = Sequential()
model.add(Dense(128, activation = "relu", input_dim=2))
model.add(Dense(128, activation = "relu"))
model.add(Dense(1, activation = "sigmoid"))
```

```python
model.summary()
```

```python
adam = Adam(learning_rate = 0.01)
model.compile(loss = 'binary_crossentropy', optimizer = adam, metrics=['accuracy'])
history = model.fit(X, Y, epochs=500, validation_split = 0.2)
```

```python
plot_decision_regions(X, Y, clf = model, legend=1)
```

```python
plt.plot(history.history['loss'], label='train')
plt.plot(history.history['val_loss'], label='test')
plt.legend()
plt.show()
```

```python
plt.plot(history.history['accuracy'], label='train')
plt.plot(history.history['val_accuracy'], label='test')
plt.legend()
plt.show()
```

#With regularization

Using L1

```python
model1 = Sequential()
model1.add(Dense(128, activation = "relu", input_dim=2, kernel_regularizer=tensorflow.keras.regularizers.l1(0.001)))
model1.add(Dense(128, activation = "relu", kernel_regularizer=tensorflow.keras.regularizers.l1(0.001)))
model1.add(Dense(1, activation = "sigmoid"))
```

```python
model1.summary()
```

```python
adam = Adam(learning_rate = 0.01)
model1.compile(loss = 'binary_crossentropy', optimizer = adam, metrics=['accuracy'])
history1 = model1.fit(X, Y, epochs=500, validation_split = 0.2)
```

```python
plot_decision_regions(X, Y, clf = model1, legend=1)
```

```python
plt.plot(history1.history['loss'], label='train')
plt.plot(history1.history['val_loss'], label='test')
plt.legend()
plt.show()
```

```python
plt.plot(history1.history['accuracy'], label='train')
plt.plot(history1.history['val_accuracy'], label='test')
plt.legend()
plt.show()
```

```python
model2 = Sequential()
model2.add(Dense(128, activation = "relu", input_dim=2, kernel_regularizer=tensorflow.keras.regularizers.l2(0.001)))
model2.add(Dense(128, activation = "relu", kernel_regularizer=tensorflow.keras.regularizers.l2(0.001)))
model2.add(Dense(1, activation = "sigmoid"))
```
